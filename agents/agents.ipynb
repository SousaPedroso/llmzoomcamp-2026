```python
def calculate_gemini35flash_price(input_tokens, output_tokens):
    INPUT_PRICE_PER_MILLION = 1.5
    OUTPUT_PRICE_PER_MILLION = 9.0

    input_cost = (input_tokens / 1_000_000) * INPUT_PRICE_PER_MILLION
    output_cost = (output_tokens / 1_000_000) * OUTPUT_PRICE_PER_MILLION
    total_cost = input_cost + output_cost

    return {
        "input_cost": input_cost,
        "output_cost": output_cost,
        "total_cost": total_cost,
    }

result = calculate_gemini35flash_price(
    token_response.total_tokens, metadata.candidates_token_count
)
print("Total cost: $", round(result["total_cost"], 8))
```

Loop Agêntico

```python
instructions = """
You're a course teaching assistant.
You're given a question from a course student and your task is to answer it.

If you want to look up information, use the search function. 
Use as many keywords from the user question as possible when making first requests.

Make multiple searches.

Try to expand your search by using new keywords
based on the results you get from the search.

At the end, ask if there are other areas that the user wants to explore.
""".strip()
```

```python
def search(query):
    boost_dict = {'question': 3.0, 'section': 0.5}
    filter_dict = {'course': 'llm-zoomcamp'}

    return index.search(
        query,
        num_results=5,
        boost_dict=boost_dict,
        filter_dict=filter_dict,
    )
```

```python
search_tool = {
    "function_declarations": [
        {
            "name": "search",
            "description": "Search the FAQ database for entries matching the given query.",
            "parameters": {
                "type": "object",
                "properties": {
                    "query": {
                        "type": "string",
                        "description": "Search query text"
                    }
                },
                "required": ["query"],
            }
        }
    ]
}
```

Define as instruções do agente

```python
config = types.GenerateContentConfig(
    system_instruction=instructions,
    tools=[search_tool],
)
```

```python
def make_call(call: types.FunctionCall) -> types.Part:
    args = call.args or {}

    if call.name == "search":
        result = search(**args)
    else:
        result = {"error": f"Unknown function: {call.name}"}

    return types.Part(
        function_response=types.FunctionResponse(
            id=call.id,
            name=call.name,
            response={"result": result},  # precisa ser dict
        )
    )
```

This readequates for a compatibility-with gemini API

```python
messages = [
    {
        'role': 'user',
        'parts': [types.Part(text="I just discovered the course. Can I join it?")
        ]
    }
]
```

```python
MAX_ITERATIONS = 10

for it in range(1, MAX_ITERATIONS + 1):
    print(f"iteration #{it}...")

    response = genai_client.models.generate_content(
        model="gemini-3.5-flash-lite",
        contents=messages,
        config=config,
    )

    model_content = response.candidates[0].content
    parts = model_content.parts or []

    # 1) guarda a fala do modelo inteira (preserva thought signatures)
    messages.append(model_content)

    # 2) separa texto e chamadas de função
    function_calls = []
    for part in parts:
        if part.function_call:
            function_calls.append(part.function_call)
        elif part.text and not part.thought:
            print("ASSISTANT:")
            print(part.text)

    # 3) sem chamadas de função = o modelo terminou
    if not function_calls:
        break

    # 4) executa todas as chamadas e devolve num único turno
    response_parts = []
    for call in function_calls:
        print(f"function_call: {call.name}({call.args})")
        response_parts.append(make_call(call))

    messages.append(types.Content(role="user", parts=response_parts))
else:
    print("Limite de iterações atingido.")
```
